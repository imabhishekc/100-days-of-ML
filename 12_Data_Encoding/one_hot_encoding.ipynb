## OneHotEncoding with Top Categories

```python
counts = df['brand'].value_counts()
```

```python
counts
```

```python
df['brand'].nunique()
```

```python
# let's take a threshold of 100 cars that will be the minimum limit to have a column of a car brand
threshold = 100
```

```python
replace_column = counts[counts <= threshold].index
```

```python
replace_column
```

```python
new_df = pd.get_dummies(df['brand'].replace(replace_column, 'others'))
```
