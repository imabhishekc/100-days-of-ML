## 1. Decide what you want to test?
Before writing code, be clear about why the dataset exists.

**Ask yourself:**
- Classification or regression?
- Linear or non-linear patterns?
- Balanced or imbalanced classes?
- Clean or noisy data?
- Categorical, numerical, or mixed?

**Examples**

- Testing encoders → categorical-heavy dataset
- Testing overfitting → very small dataset with noise
- Testing model robustness → label noise + outliers

## Use `sklearn.datasets` (fastest way)

Scikit-learn already gives you perfect ***toy generators***.

## Classification

### [Scikit-learn make_classification documentation](https://scikit-learn.org/stable/modules/generated/sklearn.datasets.make_classification.html)

```python
from sklearn.datasets import make_classification

X, y = make_classification(
    n_samples=500,
    n_features=4,
    n_informative=2,
    n_redundant=0,
    n_classes=2,
    class_sep=1.5,
    random_state=42
)
```

## Regression

### [Scikit-learn make_regression documentation](https://scikit-learn.org/stable/modules/generated/sklearn.datasets.make_regression.html)

```python
from sklearn.datasets import make_regression

X, y = make_regression(
    n_samples=300,
    n_features=3,
    noise=15,
    random_state=42
)
```

## Clustering / blobs

### [Scikit-learn make_blobs documentation](https://scikit-learn.org/stable/modules/generated/sklearn.datasets.make_blobs.html)

```python
from sklearn.datasets import make_blobs

X, y = make_blobs(
    n_samples=400,
    centers=3,
    cluster_std=1.2,
    random_state=42
)
```

## Building a dataset manually
#### Example: simple rule-based classification

```python
import numpy as np
import pandas as pd

np.random.seed(42)

n = 200
age = np.random.randint(18, 60, n)
income = np.random.randint(20, 120, n)

# rule-based label
buy = ((age < 35) & (income > 60)).astype(int)

df = pd.DataFrame({
    "age": age,
    "income": income,
    "buy_product": buy
})

df.head()
```

## Adding categorical features (for Encoding practice)

```python
city = np.random.choice(["Kathmandu", "Pokhara", "Biratnagar"], n)
membership = np.random.choice(["free", "silver", "gold"], n)

df["city"] = city
df["membership"] = membership
```

**This helps us testing:**
- LabelEncoding
- OneHotEncoding
- OrdinalEncoding

## Injecting noise & realism in our Toy Dataset.
#### (Therefore: Realworld data are noisy and messy.)

### Label noise

```python
noise_index = np.random.choice(df.index, size=20, replace=False)
df.loc[noise_index, "buy_product"] ^= 1
```

### Missing values

```python
df.loc[np.random.choice(df.index, 15), "income"] = np.nan
```

### Missing values

```python
df.loc[np.random.choice(df.index, 5), "income"] = 500
```

## Saving the final dataset as `toy_customer_data.csv`

```python
df.to_csv("toy_customer_data.csv", index=False)
```

## Visual intution of our toy dataset `toy_customer_data.csv`

```python
import matplotlib.pyplot as plt

plt.scatter(df["age"], df["income"], c=df["buy_product"])
plt.xlabel("Age")
plt.ylabel("Income")
plt.show()
```

## Advanced technique: parametric dataset generator

```python
def generate_customer_data(n=300, noise=0.1):
    age = np.random.randint(18, 60, n)
    income = np.random.randint(20, 120, n)
    y = ((age < 35) & (income > 60)).astype(int)

    flip = np.random.rand(n) < noise
    y[flip] = 1 - y[flip]

    return pd.DataFrame({
        "age": age,
        "income": income,
        "target": y
    })
```

**This lets you test:**
- Different dataset sizes
- Noise sensitivity
- Model stability

### The following table gives an Idea to adapt approach for each goal:
| Goal               | Best approach         |
| ------------------ | --------------------- |
| Quick testing      | `make_classification` |
| Learning ML deeply | Manual rule-based     |
| Encoding practice  | Mixed categorical     |
| Robustness testing | Add noise + outliers  |
| Tool benchmarking  | Parametric generator  |
