```python
import numpy as np
import pandas as pd
```

```python
def generate_nominal_dataset(n=600,noise=0.1):
    
    rng = np.random.default_rng(seed=42)

    # nominal categories
    countries = ['Germany', 'Australia', 'USA', 'UK']
    devices = ['mobile', 'desktop', 'tablet']
    payment = ['card', 'cash', 'ibanking', 'wallet']

    country = rng.choice(countries, size=n, p=[0.35, 0.25, 0.25, 0.15])
    device_type = rng.choice(devices, size=n, p=[0.6, 0.3, 0.1])
    payment_method = rng.choice(payment, size=n)

    age = rng.integers(18, 65, size=n)

    base_prob = np.full(n, 0.25)

    country_effect = {
        'Germany': 0.15,
        'Australia': 0.05,
        'USA': 0.10,
        'UK': 0.02
    } # each country has it's independent impact on data

    device_effect = {
        'mobile': 0.08,
        'desktop': 0.12,
        'tablet': 0.04,
    }

    payment_effect = {
        "card": 0.10,
        "cash": 0.02,
        "ibanking": 0.12,
        "wallet": 0.06
    }

    probablity = base_prob.copy()

    for i in range(n):
        probablity[i] += (
            country_effect[country[i]] + device_effect[device_type[i]] + payment_effect[payment_method[i]]
        )

    probablity = np.clip(probablity, 0, 1)

    purchase_complete = rng.random(n) < probablity

    noise_mask = rng.random(n) < noise
    purchase_complete[noise_mask] = ~purchase_complete[noise_mask]

    purchase_complete = np.where(purchase_complete, "yes", "no")

    return pd.DataFrame({
        'age': age,
        'country': country,
        'device_type': device_type,
        'payment_method': payment_method,
        'purchase_complete': purchase_complete,
    })
```

```python
df = generate_nominal_dataset(1000)
df.to_csv('synthetic_nominal_customer.csv', index=None)
```
