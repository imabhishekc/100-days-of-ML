```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv('ordinal_interview_dummy_dataset.csv')
```

```python
df.sample(10)
```

```python
df = df.iloc[:, [2, 3, 5]]
```

```python
df 
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(df.iloc[:,0:2],df.iloc[:,-1],test_size=0.2)
```

```python
X_train['education_level'].unique()
```

```python
X_train['job_level'].unique()
```

### OrdinalEncoder for feature columns

```python
from sklearn.preprocessing import OrdinalEncoder
encoder = OrdinalEncoder(categories=[['high_school', 'bachelor', 'master', 'phd'], ['entry', 'junior', 'mid', 'senior', 'lead']])
```

```python
encoder.fit(X_train)
```

```python
X_train = encoder.transform(X_train)
X_test = encoder.transform(X_test)
```

```python
X_train
```

```python
encoder.categories_
```

### Label encoder for target column

```python
from sklearn.preprocessing import LabelEncoder

label_encoder = LabelEncoder()
```

```python
label_encoder.fit(y_train)
```

```python
label_encoder.classes_
```

```python
y_train = label_encoder.transform(y_train)
y_test = label_encoder.transform(y_test)
```
