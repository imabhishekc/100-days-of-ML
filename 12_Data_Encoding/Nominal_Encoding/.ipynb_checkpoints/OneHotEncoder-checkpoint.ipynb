```python
df = pd.read_csv('synthetic_nominal_customer.csv')
```

### pandas OneHotEncoding

```python
pandas_one_hot_encoding = pd.get_dummies(df, columns=['country', 'device_type', 'payment_method'], prefix=None, prefix_sep='_', dummy_na=False, sparse=False, drop_first=True, dtype=None)
```

## Sci-kit learn OneHotEncoding

### Train-test splip
#### (It is a good practice to split the training and testing data)

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(df.iloc[:,0:4], df.iloc[:,-1], test_size=0.2, random_state=None)
```

```python
X_train
```

```python
y_train
```

```python
from sklearn.preprocessing import OneHotEncoder

encoder = OneHotEncoder(drop='first')
```

```python
X_train_transformed = encoder.fit_transform(X_train[['country', 'device_type', 'payment_method']]).toarray()
```

```python
X_train_transformed.shape
```

```python
X_test_transformed = encoder.fit_transform(X_test[['country', 'device_type', 'payment_method']]).toarray()
```

```python
X_test_transformed.shape
```

```python
np.hstack((X_train[['country', 'device_type', 'payment_method']], X_train_transformed))
```

```python
np.hstack((X_test[['country', 'device_type', 'payment_method']], X_test_transformed))
```
